```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_excel('/content/Supermarket_Sales_Data.xlsx')
df
```

```python
df.info()
```

```python
df.describe()
```

```python
df['Branch']=df['Branch'].str.strip().str.title()
```

```python
df['Branch'].value_counts()
```

```python
df['Customer Type']=df['Customer Type'].str.strip().str.title()
```

```python
df['Customer Type'].value_counts()
```

```python
df['Date'].value_counts()
```

```python
df['Date']=pd.to_datetime(df['Date'],format='mixed')
df['Date'] = df['Date'].dt.strftime('%Y-%m-%d')
```

```python
df.isna().sum()
```

```python
df['Quantity'].fillna(df['Quantity'].median(),inplace=True)
```

```python
df['Gross Income'].fillna(df['Gross Income'].median(),inplace=True)
```

```python
df.isna().sum()
```

```python
df['Total']=df['Unit Price']*df['Quantity']
```

```python
df['Gender']=df['Gender'].str.strip().str.title()
```

```python
df['Gender'].value_counts()
```

```python
df.columns = (df.columns.str.strip().str.lower().str.replace(r'[^a-z0-9]+', '_', regex=True))
```
